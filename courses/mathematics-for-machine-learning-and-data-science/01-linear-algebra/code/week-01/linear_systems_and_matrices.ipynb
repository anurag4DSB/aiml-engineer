equation:

-$x_1$ + 3$x_2$ = 7

3$x_1$ + 2$x_2$ = 1

solution: np.linalg.solve(A, b)
determinant: np.linalg.det(A)

```python
import numpy as np
```

```python
A = np.array([[-1, 3], [3, 2]])
b = np.array([7, 1])
solution = np.linalg.solve(A, b)
print(f'Solution: {solution}')
print(f'Shape of A/n{A.shape}')
print(f'Shape of b/n{b.shape}')
```

```python
determinant = np.linalg.det(A)
print(f"{determinant:.2f}")
# unique solution as non zero determinant
```

Visualizing 2x2 Systems as Plotlines

```python
import matplotlib.pyplot as plt


def plot_lines(system, x_range=(-10, 10)):
    A, b = system[:, :2], system[:, 2]
    x = np.linspace(*x_range, 100)
    fig, ax = plt.subplots(figsize=(6, 6))

    for (a1, a2), c in zip(A, b):
        label = f"{a1}x1 + {a2}x2 = {c}"
        if a2 != 0:
            ax.plot(x, (c - a1 * x) / a2, label=label)
        else:
            ax.axvline(c / a1, label=label)

    try:
        sol = np.linalg.solve(A, b)
        ax.plot(*sol, 'ko', label=f"solution: ({sol[0]:.2f}, {sol[1]:.2f})")
    except np.linalg.LinAlgError:
        pass

    ax.axhline(0, color='gray', linewidth=0.5)
    ax.axvline(0, color='gray', linewidth=0.5)
    ax.set_xlabel('x1')
    ax.set_ylabel('x2')
    ax.set_xlim(x_range)
    ax.legend()
    ax.grid(True)
    plt.show()
```

```python
A_system = np.hstack((A, b.reshape((2, 1))))
plot_lines(A_system)
```

Unique solution: the two lines intersect exactly at the solution point computed above.

## No solution: parallel lines

If one row of `A` is a scalar multiple of the other but `b` isn't scaled the same way, the system is inconsistent -- the lines never meet.

```python
A_parallel = np.array([[1, 2], [2, 4]])
b_parallel = np.array([3, 5])
print(f"Determinant: {np.linalg.det(A_parallel):.2f}")

try:
    np.linalg.solve(A_parallel, b_parallel)
except np.linalg.LinAlgError as err:
    print(err)
```

```python
parallel_system = np.hstack((A_parallel, b_parallel.reshape((2, 1))))
plot_lines(parallel_system)
```

As expected, the lines are parallel and never intersect -- no solution.

## Infinite solutions: coincident lines

Same `A_parallel`, but this time scale `b` to match the row proportion exactly. Now every point on the line satisfies both equations.

```python
b_coincident = np.array([3, 6])
coincident_system = np.hstack((A_parallel, b_coincident.reshape((2, 1))))
plot_lines(coincident_system)
```

The two lines overlap completely -- infinitely many solutions.
